```python
tokenizer = tf.keras.preprocessing.text.Tokenizer()

tokenizer.fit_on_texts(movie_name)
```

```python
seq = tokenizer.texts_to_sequences(movie_name)

seq[:10]
```

```python
tokenizer.word_index
```

```python
X = []
y = []

total_words_dropped = 0

for i in seq:
    if len(i) > 1:
        for index in range(1, len(i)):
            X.append(i[:index])
            y.append(i[index])
    else:
        total_words_dropped += 1

print("Total Single Words Dropped are:", total_words_dropped)
```

```python
X[:10]
```

```python
y[:10]
```

```python
X = tf.keras.preprocessing.sequence.pad_sequences(X)
```

```python
X = tf.keras.preprocessing.sequence.pad_sequences(X)

X
```

```python
X.shape
```

```python
y = tf.keras.utils.to_categorical(y)

y
```

```python
y.shape
```

```python
vocab_size = len(tokenizer.word_index) + 1

vocab_size
```

```python
vocab_array = np.array(list(tokenizer.word_index.keys()))
```

```python
vocab_array
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Embedding(vocab_size, 14),
    tf.keras.layers.LSTM(100, return_sequences=True),
    tf.keras.layers.LSTM(100),
    tf.keras.layers.Dense(100, activation='relu'),
    tf.keras.layers.Dense(vocab_size, activation='softmax'),
])
```

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
model.summary()
```

```python
model.fit(X, y, epochs=200)
```

```python
model.save('Pushpa_NextWordPrediction.h5')
```

```python
def make_prediction(text, n_words):
    for i in range(n_words):
        text_tokenize = tokenizer.texts_to_sequences([text])
        text_padded = tf.keras.preprocessing.sequence.pad_sequences(
            text_tokenize,
            maxlen=X.shape[1])
        prediction = np.squeeze(
            np.argmax(model.predict(text_padded, verbose=0), axis=-1))
        if prediction == 0:
            break
        prediction = str(vocab_array[prediction - 1])
        print(prediction)
        text += " " + prediction
    return text
```
