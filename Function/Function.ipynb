# Arguments

Arguments are the actual values passed to a function when it is called, whereas parameters are the variables defined in the function signature that receive those values

Arguments are two types:
- Formal Argument
- Actual Argument

```python
def add(x,y):  # x & y -- formal argument 
    c=x+y
    print(c)  
add(5,6) # 5 & 6 --  actual argument 
```

```python
def add(x,y):
    c = x + y
    print(c)

r = add(5,6)
print(r)
print(type(r))
```

```python
def add(x,y):
    c = x + y
    return c # The return statement in Python is used inside a function to exit the function and send a value back to the caller. 

r = add(5,6)
print(r)
print(type(r))
```

# Actual Argument types:

- Positional Argument
- Keyword Argument
- Default Argument
- Variable Argument
- keyword + Variable length Argument

# Positional Argument

```python
def person(name,age):
    print(name)
    print(age)
    
person('nit')
```

```python
def person(name):
    print(name)
    print(age)
    
person('nit', 23) 
```

```python
def person(name, age):
    print(name)
    print(age)
    
person('nit', 23) 
```

```python
def person(age, name):
    print(name)
    print(age)
    
person('nit', 23) 
```

```python
def person(age, name):
    print(name)
    print(age+1)
    
person('nit', 23)  
```

# Keyword Argument

```python
def person(age, name):
    print(name)
    print(age+1)

person('nit', 23)  
```

```python
def person(age, name):
    print(name)
    print(age+1)

person(name = 'nit', age = 23)  
```

```python
def person(age, name, mob):
    print(name)
    print(age+1)
    print(mob)

person(name = 'nit', age = 23, 12345)  
```
