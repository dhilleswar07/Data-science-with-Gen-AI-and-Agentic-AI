# Local Variables

```python
def greet():
    msg = "Hello from inside the function!"
    print(msg)

greet()
```

```python
def greet():
    msg = "Hello!"
    print("Inside function:", msg)

greet()
print("Outside function:", msg)
```

# Global Variables

```python
msg = "Python is awesome!"

def display():
    print("Inside function:", msg)

display()
print("Outside function:", msg)
```

```python
def fun():
    s = "I Love Coding."
    print(s)

s = "Me Too"
fun()   
print(s)
```

```python
def fun():
    s += ' Dhilleswar'   # Error: Python thinks s is local
    print(s)

s = "vijay"
fun()
```

```python
s = "Python is great!"

def fun():
    global s
    s += " Dhilleswar"   # Modify global variable
    print(s)
    s = "Vijay"  # Reassign global
    print(s)

fun()
print(s)
```

```python
a = 1  # Global variable

def f():
    print("f():", a)  # Uses global a

def g():
    a = 2  # Local shadows global
    print("g():", a)

def h():
    global a
    a = 3  # Modifies global a
    print("h():", a)

print("global:", a)
f()
print("global:", a)
g()
print("global:", a)
h()
print("global:", a)
```
